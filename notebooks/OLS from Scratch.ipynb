### **$y$ = 3 + 2 $x$ + $ϵ$**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(45)
X = np.linspace(0, 5, 500)
noise = np.random.normal(loc=0, scale=1, size=X.shape)

y = 2*X + 3 + noise
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(X, y)
plt.xlabel("X")
plt.ylabel("y")
plt.title("Synthetic Linear Regression Dataset")

plt.show()
```

```python
X_design = np.column_stack((np.ones(len(X)),X))
XtX = X_design.T @ X_design
Xty = X_design.T @ y

beta = np.linalg.solve(XtX, Xty)
print("Estimated coefficients:", beta)
```

```python
y_pred = X_design @ beta

plt.figure(figsize=(8, 5))
plt.scatter(X, y, label="Data")
plt.plot(X,y_pred,label="OLS prediction", color="red")
plt.xlabel("X")
plt.ylabel("y")
plt.title("OLS Regression from Scratch")
plt.legend()

plt.show()
```

```python
residuals = y - y_pred
mse = np.mean(residuals**2)
print("Mean Squared Error:", mse)
rmse = np.sqrt(mse)
print("Root Mean Squared Error:", rmse)
```

```python
y_mean = np.mean(y)
SSE = np.sum((y - y_pred)**2)
SST = np.sum((y - y_mean)**2)

r_2 = 1 - (SSE/SST)
print("R-squared:", r_2)
```
